Шпаргалка по регулярным выражения

\. - любой символ

\d или [0-9] - любой цифровой символ

\D или [^0-9] - любой не цифровой символ

\s или [ \f\n\r\t\v] - пробельный символ

\S или [^ \f\n\r\t\v] - не пробельный символ

\w или [A-Za-z0-9_] - буквенный, цифровой или знак подчеркивания

\W или [^A-Za-z0-9_] - любой кроме буквенного, цифрового и знака подчеркивания

? или {0, 1} - предыдущий символ может встречаться ноль или один раз

\* или {0, } - предыдущий символ может встречаться ноль и более раз

\+ или {1, } - предыдущий символ может встреться один или более раз

{n} - предыдущий символ встречается ровно n раз

{m, n} - предыдущий символ встречается от n до m раз

{m, } -предыдущий символ встречается не менее m раз

{, n} - предыдущий символ встреччается не более n раз

Жадный оператор выбирает максимально длинную подстроку из возможных

Ленивый оператор выбирает максимально короткую подстроку из возможных

жадные операторы:

\*

\+

{n, }

ленивые операторы:

\*?

\+?

{n, }?

| - или

[ ] - диапазон

Список служебных символов (необходимо экранировать): [, ], \\, |\, /\, ^, $, ., ?, \*, \+, (, ),{, }

### Удаление стоп-слов

```python
# функция удаляет стоп слова
def delete_stopwords(texts: List[List[str]], stop_words: List[str]) -> List[List[str]]:
    result = []
    for words in texts:
        clear_words = []
        for word in words:
            if word not in stop_words:
                clear_words.append(word)
        result.append(clear_words)
    return result
```

TF-IDF (Term Frequency – Inverse Document Frequency) — это способ представить тексты в виде чисел, чтобы с ними могла работать модель.

Каждое слово → число.
Каждый текст → вектор чисел.

**Как выглядит TF-IDF матрица?**
Это обычная таблица (матрица), где:

Строки — это тексты (документы).

Столбцы — это уникальные слова (термы) после токенизации.

Значения — это "вес" каждого слова в каждом тексте.

🔢 Пример (на 3 документа и 5 слов):




💡 TF-IDF = TF × IDF
Это комбинация двух частей:
1. TF (Term Frequency) — насколько часто слово встречается в тексте
→ Считается как:
кол-во раз слово встречается в тексте ÷ общее кол-во слов в тексте.

Например, текст:

"банк банк карта"

TF("банк") = 2 / 3 = 0.67

TF("карта") = 1 / 3 = 0.33


2. IDF (Inverse Document Frequency) — насколько редкое слово в целом по всем документам
→ Формула:
log(общее кол-во документов ÷ кол-во документов, где слово встречается)

Пример:
Если у нас 1000 отзывов, и слово "банк" встречается в 900 из них, то:
IDF("банк") = log(1000 / 900) = log(1.11) ≈ 0.045 → маленькое число

А если слово "обман" встречается только в 10 документах:
IDF("обман") = log(1000 / 10) = log(100) ≈ 2.0 → большое число

🔁 А теперь вместе: TF × IDF
Если слово часто встречается в документе (TF высокий), и редкое в других документах (IDF высокий)
→ оно получает высокий вес в этом документе.

Если слово часто везде (например "банк", "карта")
→ его вес низкий, потому что оно ничего не говорит о сути конкретного отзыва.



```python
nltk.download('punkt') # 	Алгоритмы для токенизации (разделения текста на предложения и слова)
nltk.download('stopwords')

class Preprocess_bad:
    def __init__(self, texts):
        self.data = texts

    def tf_idf(self):
        vectorizer = TfidfVectorizer()
        tfidf_matrix = vectorizer.fit_transform(self.data)
        return tfidf_matrix, vectorizer

    
    def w2v(self, vector_size=100, window=5, min_count=2, workers=4, epochs=5):
        tokenized = [doc.split() for doc in self.data]
        model = Word2Vec(
            sentences=tokenized,
            vector_size=vector_size,
            window=window,
            min_count=min_count,
            workers=workers,
            epochs=epochs
        )
        doc_vectors = []
        for tokens in tokenized:
            vectors = [model.wv[word] for word in tokens if word in model.wv]
            if vectors:
                doc_vectors.append(np.mean(vectors, axis=0))
            else:
                doc_vectors.append(np.zeros(vector_size))

        return np.array(doc_vectors)

class Preprocess_good:
    def __init__(self):
        self.tfidf_vectorizer = TfidfVectorizer(min_df=5, max_df=0.8)  #min_df - сколько минимально надо встретиться слову, чтобу попасть в матрицу. Max_df - % документов, елси слово содержится больше чем в нем, то игнорится
        self.w2v_model = None
        self.mystem = Mystem()
        self.russian_stopwords = set(stopwords.words("russian"))

    def clean_html(self, text):
        return re.sub(r'<.*?>', '', text)

    def token(self, text):
        text = self.clean_html(text)
        text = re.sub(r'[^а-яa-zё]+', ' ', text.lower())
        return text.split()

    def stop_words(self, tokens):
        return [token for token in tokens if token not in self.russian_stopwords]

    def lemmatization(self, tokens):
        text = ' '.join(tokens)
        lemmas = self.mystem.lemmatize(text)
        return [lemma for lemma in lemmas if lemma.strip() and lemma not in punctuation]

    def preprocess_text(self, text):
        tokens = self.token(text)
        tokens = self.stop_words(tokens)
        lemmas = self.lemmatization(tokens)
        return lemmas

    def tf_idf(self, texts):
        processed_texts = [' '.join(self.preprocess_text(text)) for text in texts]
        tfidf_matrix = self.tfidf_vectorizer.fit_transform(processed_texts)
        return tfidf_matrix

    def w2v(self, texts, vector_size=100, window=5, min_count=2, workers=4, epochs=5): # vector_size-размерность веторка, window-размер окна контекста, workers-потоки обучения, epochs-число циклов
        tokenized_texts = [self.preprocess_text(text) for text in texts]
        self.w2v_model = Word2Vec(
            sentences=tokenized_texts,
            vector_size=vector_size,
            window=window,
            min_count=min_count,
            workers=workers,
            epochs=epochs
        )

        doc_vectors = []
        for tokens in tokenized_texts:
            vectors = [self.w2v_model.wv[word] for word in tokens if word in self.w2v_model.wv]
            if vectors:
                doc_vectors.append(np.mean(vectors, axis=0))
            else:
                doc_vectors.append(np.zeros(vector_size))

        return np.array(doc_vectors)
```
