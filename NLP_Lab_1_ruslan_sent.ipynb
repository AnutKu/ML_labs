### Регулярные выражения

Шпаргалка по регулярным выражения

\. - любой символ

\d или [0-9] - любой цифровой символ

\D или [^0-9] - любой не цифровой символ

\s или [ \f\n\r\t\v] - пробельный символ

\S или [^ \f\n\r\t\v] - не пробельный символ

\w или [A-Za-z0-9_] - буквенный, цифровой или знак подчеркивания

\W или [^A-Za-z0-9_] - любой кроме буквенного, цифрового и знака подчеркивания

? или {0, 1} - предыдущий символ может встречаться ноль или один раз

\* или {0, } - предыдущий символ может встречаться ноль и более раз

\+ или {1, } - предыдущий символ может встреться один или более раз

{n} - предыдущий символ встречается ровно n раз

{m, n} - предыдущий символ встречается от n до m раз

{m, } -предыдущий символ встречается не менее m раз

{, n} - предыдущий символ встреччается не более n раз

Жадный оператор выбирает максимально длинную подстроку из возможных

Ленивый оператор выбирает максимально короткую подстроку из возможных

жадные операторы:

\*

\+

{n, }

ленивые операторы:

\*?

\+?

{n, }?

| - или

[ ] - диапазон

Список служебных символов (необходимо экранировать): [, ], \\, |\, /\, ^, $, ., ?, \*, \+, (, ),{, }

### Лемматизация и стемминг

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
# nltk.download('wordnet')

def compare_stemmer_and_lemmatizer(stemmer, lemmatizer, word):
    """
    Print the results of stemmind and lemmitization using the passed stemmer, lemmatizer, word and pos (part of speech)
    """
    print("Stemmer:", stemmer.stem(word))
    print("Lemmatizer:", lemmatizer.lemmatize(word, pos='v'))
    print()

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()
compare_stemmer_and_lemmatizer(stemmer, lemmatizer, word = "seen")
compare_stemmer_and_lemmatizer(stemmer, lemmatizer, word = "drove")
```

### Удаление стоп-слов

```python
# функция удаляет стоп слова
def delete_stopwords(texts: List[List[str]], stop_words: List[str]) -> List[List[str]]:
    result = []
    for words in texts:
        clear_words = []
        for word in words:
            if word not in stop_words:
                clear_words.append(word)
        result.append(clear_words)
    return result
```

```python
from gensim.models.word2vec import Word2Vec  # подключение модели word2vec
from tqdm.notebook import tqdm  # подключения библиотеки прогресс бара
```

```python
def get_wv_data(data):
    wv_data = []
    for s in tqdm(data):
        lst = re.findall("[a-z]+", s.lower())
        wv_data.append(lst)
    return wv_data
```

```python
wv_data = get_wv_data(df.text.to_list())
```

```python
wv_data
```

```python
wv_model = Word2Vec(vector_size=100, window=5, min_count=10, negative=10, seed=42)  # инициализация модели
wv_model.build_vocab(wv_data)  # сбор словаря
```

```python
wv_model.train(wv_data, total_examples=wv_model.corpus_count, epochs=5) # Обучение модели
```
