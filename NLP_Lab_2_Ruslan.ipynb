```python
# подключение библиотек
import re
import pandas as pd  # библиотека для работы с табличными данными
import numpy as np

df = pd.read_csv("data.csv.gz", compression='gzip') # чтение таблицы
```

### Word2Vec

```python
from gensim.models.word2vec import Word2Vec  # подключение модели word2vec
from tqdm.notebook import tqdm  # подключения библиотеки прогресс бара

def get_wv_data(data):
    wv_data = []
    for s in tqdm(data):
        lst = re.findall("[а-яё]+", s.lower())
        wv_data.append(lst)
    return wv_data
```

```python
X_train_wv = get_wv_data(X_train)
X_valid_wv = get_wv_data(X_valid)
```

```python
# в зависимости от версии библиотеки параметр отвечающий за размер вектора может называться либо size либо vector_size

wv_model = Word2Vec(vector_size=100, window=5, min_count=10, negative=10, seed=42)  # инициализация модели
wv_model.build_vocab(X_train_wv)  # сбор словаря
```

```python
wv_model.train(X_train_wv, total_examples=wv_model.corpus_count, epochs=5)  # обучение модели
```

```python
from typing import List

# функция формирует из списка слов вектор
def create_text_vector(words: List[str], wv_model: Word2Vec) -> np.array:
    vectors = []
    for word in words:
        if wv_model.wv.has_index_for(word):
            vector = wv_model.wv.get_vector(word)
            vectors.append(vector)
    # число векторов равно 0, берем средний вектор всей коллекции
    if len(vectors) == 0:
        vectors = wv_model.wv.vectors
    else:
        vectors = np.array(vectors)
    return vectors.mean(axis=0)

def create_vectors(data: List[List[str]], wv_model: Word2Vec) -> np.array:
    vectors = []
    for words in tqdm(data):
        vector = create_text_vector(words, wv_model)
        vectors.append(vector)
    vectors = np.array(vectors)
    return vectors
```

```python
X_train_wv_vectors = create_vectors(X_train_wv, wv_model)
```

```python
X_train_wv_vectors = create_vectors(X_train_wv, wv_model)
X_valid_wv_vectors = create_vectors(X_valid_wv, wv_model)
```

```python
import pymorphy2  # библиотека для лемматизации
import re         # библиотека для регулярных выражений
import nltk       # будем использовать для скачивая стоп слов
from typing import List  # для отражения типа данных список
from tqdm.notebook import tqdm  # прогресс бар
from nltk.stem import SnowballStemmer  # класс осуществляющий стемминг
from nltk.corpus import stopwords  # стоп слова
nltk.download('stopwords')


# функция реализующая лемматизацию
def lemmatization(texts: List[List[str]]) -> List[List[str]]:
    result = []
    d = {}
    morph_analyzer = pymorphy2.MorphAnalyzer()
    for text in tqdm(texts):
        words_lemmatized = []
        for word in text:
            if word in d:
                word_lemmatized = d[word]
            else:
                word_lemmatized = morph_analyzer.parse(word)[0].normal_form
                d[word] = word_lemmatized

            words_lemmatized.append(word_lemmatized)
        result.append(words_lemmatized)
    return result

# функция выполняет обработку регулярным выражением и приводит к нижнем регистру
def preprocessing(texts: List[str], reg_expression: str="[а-яё]+", lower: bool=True) -> List[List[str]]:
    result = []
    for text in tqdm(texts):
        if lower:
            text = text.lower()
        processed = re.findall(reg_expression, text)
        result.append(processed)
    return result

# функция удаляет стоп слова
def delete_stopwords(texts: List[List[str]], stop_words: List[str]) -> List[List[str]]:
    result = []
    for words in texts:
        clear_words = []
        for word in words:
            if word not in stop_words:
                clear_words.append(word)
        result.append(clear_words)
    return result

# функция преобразует список списков слов в список строк
def list2str(texts: List[List[str]]) -> List[str]:
    result = []
    for text in texts:
        string = " ".join(text)
        result.append(string)
    return result
```

```python
import gensim.corpora as corpora

# Create Dictionary
id2word = corpora.Dictionary(X_lem_without_stopwords)

# Create Corpus: a collection of documents and each document is a list of words
texts = X_lem_without_stopwords

# Convert each document into a bag-of-words (BoW) format = a list of (token_id, token_count) tuples.
corpus = [id2word.doc2bow(text) for text in texts]

print(corpus[0])
print(texts[0])
```

```python
print(corpus[100])
print(texts[100])
```

```python
# number of topics
K = 5

# Build LDA model
lda_model = LdaModel(corpus=corpus,
                                       id2word=id2word,
                                       num_topics=K,alpha='asymmetric',eta=0.01)
```

```python
for idx,topic in lda_model.print_topics():
    print("Topic: {} \nWords: {}".format(idx, topic))
    print("\n")
```

```python
lda_model.print_topics()
```

```python
doc_lda = lda_model[corpus]

# iterate through documents/titles
for i in range(len(texts)):
    docbow = corpus[i] # get the document BoW
    doc_topics = lda_model.get_document_topics(docbow,minimum_probability=0.0) # Return topic distribution for the given document BoW, as a list of (topic_id, topic_probability) 2-tuples.
    doc_topic_prob = np.array([v[1] for v in doc_topics]) # convert it into a numpy array
    if(i < 5):
        print('paper:',i+1,', dominant topic:',np.argmax(doc_topic_prob)+1) # choose the dominate topic (i.e., with the maximum topic probability among K topics)
        print('paper:',i+1,', dominant topic:', doc_topic_prob)
print(".....")
```

```python
lsa_model = TruncatedSVD(n_components=3, algorithm='randomized', n_iter=10, random_state=42)
lsa_top=lsa_model.fit_transform(values)

# Темы и слова
print("Темы:", lsa_model.components_)
print("Слова:", tf_idfvectorizer.get_feature_names())
```
